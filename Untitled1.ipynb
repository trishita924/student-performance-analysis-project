```python
data['Hours_Studied'].corr(data['Exam_Score'])
```

```python
import matplotlib.pyplot as plt

plt.scatter(data['Hours_Studied'], data['Exam_Score'])

plt.xlabel('Hours Studied')
plt.ylabel('Exam Score')
plt.title('Hours Studied vs Exam Score')

plt.show()
```

```python
data.groupby('Motivation_Level')['Exam_Score'].mean().sort_values(ascending=False)
```

```python
motivation_scores = data.groupby('Motivation_Level')['Exam_Score'].mean().sort_values(ascending=False)

motivation_scores.plot(kind='bar')

plt.xlabel('Motivation Level')
plt.ylabel('Average Exam Score')
plt.title('Motivation Level vs Average Exam Score')
plt.ylim(60, 70)

plt.show()
```

```python
data.isnull().sum()
```

```python
data['Teacher_Quality'] = data['Teacher_Quality'].fillna(
    data['Teacher_Quality'].mode()[0]
)

data['Parental_Education_Level'] = data['Parental_Education_Level'].fillna(
    data['Parental_Education_Level'].mode()[0]
)

data['Distance_from_Home'] = data['Distance_from_Home'].fillna(
    data['Distance_from_Home'].mode()[0]
)
```

```python
data.isnull().sum()
```

```python
correlation = data.corr(numeric_only=True)

correlation['Exam_Score'].sort_values(ascending=False)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Matrix of Numerical Features')
plt.show()
```
